```python
pq("""
SELECT
    customer_id,
    COUNT(*) AS total_orders,
    SUM(CASE WHEN amount >= 500 THEN 1 ELSE 0 END) AS orders_500_or_more
FROM shop_orders
GROUP BY customer_id
ORDER BY customer_id;
""")
```

```python
pq(""" 
SELECT COUNT(*) AS orders,
SUM(amount>=500) as big_orders,
ROUND(100.0*avg(amount>=500),2) as perc
From shop_orders;""")
```

```python
pq("""
SELECT
    name,
    joined_on,
    CAST(julianday('2025-03-31') - julianday(joined_on) AS INTEGER) AS days_with_shop
FROM shop_customers;
""")
```

```python
pq("""

SELECt strftime('%Y', joined_on) AS year
FROM shop_customers;
""")
```

```python
pq("""
SELECT
    strftime('%Y-%m', order_date) AS month,
    COUNT(*) AS orders,
    SUM(amount) AS sales
FROM shop_orders
GROUP BY month
ORDER BY month;
""")
```

```python
def run_sql_file(path):
    with open(path, 'r') as f:
        script = f.read()

    pcon.executescript(script)
    pcon.commit()
    print("ran", path)
```
